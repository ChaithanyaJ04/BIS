```python
import random

# --- PARAMETERS ---
POP_SIZE = 5       # Set to 5
CHROM_LEN = 5      # 5 bits allows integers from 0 to 31
GENERATIONS = 30
P_C = 0.8
P_M = 0.1

# --- HELPER: BINARY TO INTEGER ---
def binary_to_int(chrom):
    """Converts a binary list [1, 0, 1] into its base-10 integer representation"""
    return sum(bit * (2 ** idx) for idx, bit in enumerate(reversed(chrom)))

# --- FITNESS FUNCTION ---
def evaluate_fitness(chrom):
    x = binary_to_int(chrom)
    return x ** 2  # f(x) = x^2

# --- GA OPERATORS ---
def roulette_select(pop, fitnesses):
    total_fitness = sum(fitnesses)
    if total_fitness == 0:
        return list(random.choice(pop))

    r = random.uniform(0, total_fitness)
    cumulative_sum = 0
    for ind, fit in zip(pop, fitnesses):
        cumulative_sum += fit
        if cumulative_sum >= r:
            return list(ind)
    return list(pop[-1])

def crossover(p1, p2):
    if random.random() > P_C:
        return p1.copy(), p2.copy()
    k = random.randint(1, CHROM_LEN - 1)
    return p1[:k] + p2[k:], p2[:k] + p1[k:]

def mutate(chrom):
    for i in range(CHROM_LEN):
        if random.random() < P_M:
            chrom[i] = 1 - chrom[i]
    return chrom

# --- MAIN EXECUTION ---
# Initialize 5 random individuals
pop = [[random.randint(0, 1) for _ in range(CHROM_LEN)] for _ in range(POP_SIZE)]

for gen in range(GENERATIONS):
    fitnesses = [evaluate_fitness(ind) for ind in pop]

    # Track the best individual in the generation
    best_fit = max(fitnesses)
    best_ind = pop[fitnesses.index(best_fit)]
    best_x = binary_to_int(best_ind)

    print(f"Gen {gen:2d}: Best x = {best_x:2d} | Fitness (x^2) = {best_fit:4d} | Chrom = {best_ind}")

    # Terminate early if we hit the global maximum (31^2 = 961)
    if best_x == 31:
        print("Optimal solution found early!")
        break

    new_pop = []
    while len(new_pop) < POP_SIZE:
        p1 = roulette_select(pop, fitnesses)
        p2 = roulette_select(pop, fitnesses)

        c1, c2 = crossover(p1, p2)

        new_pop.append(mutate(c1))
        if len(new_pop) < POP_SIZE:
            new_pop.append(mutate(c2))

    pop = new_pop
```
