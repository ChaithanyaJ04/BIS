```python

import random

# Fixed PSO parameters
w = 0.3
c1 = 2
c2 = 2

# User input
n = int(input("Enter number of particles: "))
it = int(input("Enter number of iterations: "))

random.seed(42)

# Initialize particle positions
pos = [[random.uniform(-10, 10),
        random.uniform(-10, 10)] for i in range(n)]

# Initialize velocities
vel = [[0, 0] for i in range(n)]

# Initialize personal best
pbest = [p[:] for p in pos]

# Fitness function (De Jong function)
def fitness(p):
    return p[0]**2 + p[1]**2

# Initialize global best
gbest = min(pbest, key=fitness)[:]

# PSO algorithm
for t in range(it):

    for i in range(n):

        for d in range(2):

            r1 = random.random()
            r2 = random.random()

            # Velocity update formula
            vel[i][d] = (
                w * vel[i][d]
                + c1 * r1 * (pbest[i][d] - pos[i][d])
                + c2 * r2 * (gbest[d] - pos[i][d])
            )

            # Position update formula
            pos[i][d] = pos[i][d] + vel[i][d]

            # Limit position
            pos[i][d] = max(-10, min(10, pos[i][d]))

        # Update personal best
        if fitness(pos[i]) < fitness(pbest[i]):
            pbest[i] = pos[i][:]

    # Update global best
    gbest = min(pbest, key=fitness)[:]

# Display result
print("\nBest Position:", gbest)
print("Minimum Fitness:", fitness(gbest))

```
