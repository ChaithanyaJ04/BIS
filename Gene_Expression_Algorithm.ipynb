```python
import random

# Get input from user
n = int(input("Enter population size: "))
length = int(input("Enter chromosome length: "))
mr = float(input("Enter mutation rate: "))
cr = float(input("Enter crossover rate: "))
gen = int(input("Enter number of generations: "))

random.seed(42)

# Generate initial population
population = [[random.randint(0, 1) for _ in range(length)] for _ in range(n)]

# Gene Expression
def decode(ch):
    return int("".join(map(str, ch)), 2)

# Fitness function
def fitness(ch):
    x = decode(ch)
    return x * x

# Selection
def selection(pop):
    a, b = random.sample(pop, 2)
    return a[:] if fitness(a) > fitness(b) else b[:]

# Genetic algorithm
for g in range(gen):

    best = max(population, key=fitness)
    new_population = [best[:]]

    while len(new_population) < n:

        p1 = selection(population)
        p2 = selection(population)

        c1, c2 = p1[:], p2[:]

        # Crossover
        if random.random() < cr:
            point = random.randint(1, length - 1)
            c1 = p1[:point] + p2[point:]
            c2 = p2[:point] + p1[point:]

        # Mutation
        for child in [c1, c2]:
            for i in range(length):
                if random.random() < mr:
                    child[i] = 1 - child[i]

            if len(new_population) < n:
                new_population.append(child)

    population = new_population

# Find best solution
best = max(population, key=fitness)

print("\nBest Chromosome:", "".join(map(str, best)))
print("Best Solution:", decode(best))
print("Maximum Fitness:", fitness(best))
```
