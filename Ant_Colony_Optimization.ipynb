```python

import random
import math

# City coordinates
cities = [(0, 0), (2, 6), (5, 2), (6, 7), (8, 3)]
N = len(cities)

# ACO parameters
M = 10
Tmax = 50
alpha = 1
beta = 2
rho = 0.5
q = 100

random.seed(42)

# Calculate distance and heuristic information
D = [[math.dist(cities[i], cities[j]) for j in range(N)]
     for i in range(N)]

eta = [[1 / D[i][j] if D[i][j] > 0 else 0
        for j in range(N)] for i in range(N)]

# Initialize pheromones
tau = [[1.0] * N for _ in range(N)]

best_tour = None
best_length = float('inf')

# ACO algorithm
for t in range(Tmax):
    all_tours = []
    all_lengths = []

    # Each ant constructs a tour
    for k in range(M):
        curr = random.randrange(N)
        visited = [curr]

        while len(visited) < N:
            J = [j for j in range(N) if j not in visited]

            weights = [(tau[curr][j] ** alpha) *
                       (eta[curr][j] ** beta) for j in J]

            next_city = random.choices(
                J, weights=weights, k=1
            )[0]

            visited.append(next_city)
            curr = next_city

        # Calculate total distance including return
        L = sum(D[visited[i]][visited[(i + 1) % N]]
                for i in range(N))

        all_tours.append(visited)
        all_lengths.append(L)

        if L < best_length:
            best_length = L
            best_tour = visited[:]

    # Evaporate pheromones
    for i in range(N):
        for j in range(N):
            tau[i][j] *= (1 - rho)

    # Deposit pheromones
    for tour, L in zip(all_tours, all_lengths):
        delta = q / L

        for i in range(N):
            u = tour[i]
            v = tour[(i + 1) % N]

            tau[u][v] += delta
            tau[v][u] += delta

# Display best solution
route = [city + 1 for city in best_tour]
route.append(route[0])

print("Best Tour:", route)
print("Shortest Distance:", round(best_length, 2))

```
